# Getting started

## Modules 

```python
import os
from pyfileindex import PyFileIndex
```

## User defined functions 

```python
def touch(fname, times=None):
    with open(fname, "a"):
        os.utime(fname, times)
```

```python
def filter_function(file_name):
    return ".txt" in file_name
```

# Demonstrate PyFileIndex

## Initialise PyFileIndex

```python
pfi = PyFileIndex(path=".", filter_function=filter_function, watch=True, debug=True)
pfi
```

## Update PyFileIndex

```python
pfi.update()
pfi
```

## New directory 

```python
os.makedirs("bla")
```

```python
pfi.update()
pfi
```

## New sub directory 

```python
os.makedirs("bla/bla")
```

```python
pfi.update()
pfi
```

## New file

```python
touch("bla/bla/test.txt")
```

```python
pfi.update()
pfi
```

## Another new file 

```python
touch("bla/test.txt")
```

```python
pfi.update()
pfi
```

## Touch an existing file 

```python
touch("bla/bla/test.txt", (1330712280, 1330712292))
```

```python
pfi.update()
pfi
```

## Remove a file 

```python
os.remove("bla/bla/test.txt")
```

```python
pfi.update()
pfi
```

## Remove a directory

```python
os.rmdir("bla/bla")
```

```python
pfi.update()
pfi
```

## Clean up 

```python
os.remove("bla/test.txt")
```

```python
os.rmdir("bla")
```

```python
pfi.update()
pfi
```
